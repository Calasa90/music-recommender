## FMA Dataset | Exploración y evaluación

### 1. Objetivo

Analizar el dataset Free Music Archive (FMA) para evaluar su utilidad en el proyecto **Music Recommender | SQL Edition**.

Se busca comprobar la calidad de los datos, la disponibilidad de información musical y su potencial para construir una base de datos relacional.

### 2. Archivos analizados

Se han utilizado tres archivos del conjunto FMA:

- `tracks.csv`: metadatos de canciones, artistas y álbumes.
- `genres.csv`: catálogo de géneros musicales.
- `echonest.csv`: características de audio y métricas adicionales.

La exploración se ha realizado con Python y Pandas.

### 3. Exploración de tracks.csv

**Dimensiones:**
- 106.574 registros.
- 52 columnas.
- 106.574 identificadores `track_id` únicos.
- 0 identificadores duplicados.

Las columnas están organizadas mediante un MultiIndex en cuatro grupos:

| Grupo | Columnas |
|---|---:|
| album | 13 |
| artist | 17 |
| set | 2 |
| track | 20 |

Esta organización permite identificar entidades útiles para SQL, especialmente artistas, álbumes y canciones.

### 4. Calidad de los datos

Se han identificado 27 columnas con valores nulos.

Entre los campos más afectados destacan:

| Campo | Valores nulos |
|---|---:|
| track.lyricist | 106.263 |
| track.composer | 102.904 |
| track.language_code | 91.550 |
| track.genre_top | 56.976 |
| album.title | 1.025 |
| track.title | 1 |

Aunque algunos metadatos están muy incompletos, los títulos de las canciones presentan una cobertura prácticamente total.

El elevado número de nulos en `genre_top` requiere especial atención para un recomendador basado en géneros.

### 5. Análisis de géneros musicales

Se han examinado tres campos:

- `genre_top`: género principal.
- `genres`: identificadores de géneros asociados.
- `genres_all`: identificadores que incluyen géneros de niveles superiores.

**Resultados:**

- Canciones con géneros asociados: 104.343 (97,9 %).
- Canciones sin géneros asociados: 2.231 (2,1 %).

Por tanto, los valores nulos de `genre_top` no implican necesariamente ausencia de información musical.

El archivo `genres.csv` contiene:

- 163 géneros registrados.
- 163 identificadores únicos.
- 0 identificadores duplicados.

También incorpora el campo `parent`, que permite representar relaciones jerárquicas entre géneros y subgéneros.

Estos datos son adecuados para plantear una relación muchos a muchos entre canciones y géneros.

### 6. Exploración de Echonest

**Dimensiones:**

- 13.129 registros.
- 249 columnas.

Se han identificado los siguientes grupos de información:

- `audio_features`
- `metadata`
- `ranks`
- `social_features`
- `temporal_features`

El grupo `audio_features` contiene ocho características relevantes:

1. Acousticness
2. Danceability
3. Energy
4. Instrumentalness
5. Liveness
6. Speechiness
7. Tempo
8. Valence

Estas variables permitirían desarrollar recomendaciones basadas en similitud musical.

Sin embargo, Echonest solo contiene registros equivalentes aproximadamente al 12,3 % del catálogo FMA.

### 7. Compatibilidad entre FMA y Echonest

Se ha comprobado la correspondencia de identificadores entre ambos archivos.

**Resultados:**

- Registros de Echonest: 13.129.
- Identificadores presentes en FMA: 13.129.
- Registros sin coincidencia: 0.

Todos los registros de Echonest tienen un identificador presente en FMA.

Esto permite plantear una unión entre los metadatos musicales y las características de audio utilizando `track_id`.

### 8. Comparación inicial con Spotify Tracks 114K

| Característica | Spotify 114K | FMA |
|---|---|---|
| Registros | 114.000 | 106.574 |
| Identificadores únicos | 89.741 | 106.574 |
| Columnas principales | 21 | 52 |
| Géneros | 114 etiquetas | 163 géneros catalogados |
| Características de audio | Disponibles en el CSV principal | Disponibles parcialmente mediante Echonest |
| Estructura relacional | Requiere transformación | Metadatos diferenciados por entidades |

Spotify 114K presenta ventajas para desarrollar recomendaciones por características musicales sobre un catálogo más amplio.

FMA ofrece una estructura especialmente interesante para trabajar relaciones entre artistas, álbumes, canciones y géneros.

### 9. Posible aplicación en SQL

FMA permitiría diseñar las siguientes tablas:

- `artists`: información de artistas.
- `albums`: información de álbumes.
- `tracks`: catálogo de canciones.
- `genres`: catálogo de géneros.
- `track_genres`: tabla intermedia para relacionar canciones y géneros.
- `audio_features`: características musicales disponibles mediante Echonest.

Este diseño es una propuesta inicial y deberá revisarse durante la fase de modelado relacional.

### 10. Conclusiones

**Ventajas de FMA:**

- Identificadores únicos en el catálogo principal.
- Amplia cobertura de géneros musicales.
- Información diferenciada sobre artistas, álbumes y canciones.
- Catálogo de géneros con estructura jerárquica.
- Compatibilidad de identificadores con Echonest.
- Buen potencial para practicar relaciones SQL.

**Limitaciones:**

- Numerosos valores nulos en determinados metadatos.
- Características de audio disponibles solo para una parte del catálogo.
- Necesidad de transformar los campos de géneros antes de importarlos a SQL.
- Pendiente evaluar si el catálogo de artistas y canciones encaja con el objetivo del recomendador.

**Valoración final:**

FMA es un candidato sólido para construir y practicar una base de datos musical relacional.

Spotify Tracks 114K continúa siendo un candidato especialmente atractivo para desarrollar recomendaciones basadas en características de audio.
